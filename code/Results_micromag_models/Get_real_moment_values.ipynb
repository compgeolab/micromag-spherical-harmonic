```python
import xarray as xr
import pathlib
import pandas as pd
```

```python

# Directories
grid_dir = pathlib.Path("../../data/micromag_models/tecfiles/grids")
grid_files = sorted(grid_dir.glob("*.nc"))
moments_csv = pathlib.Path("../../data/micromag_models/real_micromagnetic_moments/dipole_moments.csv")


dipole_moments_list = []

for file in grid_files:
    parts = file.stem.split("_")  # e.g., ["grid", "eps0", "0.05"]
    model_name = parts[1]
    height = float(parts[2])

    # Filtra apenas arquivos com height = 0.05 µm
    if height != 0.05:
        continue

    print(f"Reading dipole moment for: {model_name} at height {height} µm")

    data = xr.load_dataset(file)
    moment = data.magnetic_moment.values  # shape: (3,)
    
    dipole_moments_list.append({
        "name": model_name,
        "mx": moment[0],
        "my": moment[1],
        "mz": moment[2],
    })

# Salvar como CSV
df = pd.DataFrame(dipole_moments_list)
df.to_csv(moments_csv, index=False)
```
